```python
from langchain_openai import ChatOpenAI
import requests
from langchain.agents import create_react_agent, AgentExecutor
from langchain import hub
from langchain.tools import DuckDuckGoSearchRun
from langchain.tools import tool
import json
```

```python
@tool
def get_weather_data(city: str) -> str:
    """Get the current weather data for a given city."""
    url = f"https://api.weatherstack.com/current?access_key=YOUR_ACCESS_KEY&query={city}"
    response =requests.get(url)
    return response.json()
```

```python
search_tool = DuckDuckGoSearchRun()
```

```python
llm = ChatOpenAI()
```

```python
prompt = hub.pull("hwchase17/react")
```

```python
agent = create_react_agent(
  llm = llm,
  tools = [search_tool,get_weather_data],
  prompt = prompt
)
```

```python
agent_executor = AgentExecutor(
  agent = agent,
  tools = [search_tool,get_weather_data],
  verbose = True
)
```

```python
response = agent_executor.invoke({"input": "give me the weather of azad jammu kashmir and also tell me who is the current prime minister of azad jammu kashmir?"})
print(response)
```
